```python
import numpy as np
import matplotlib.pyplot as plt

# number of terms
N = int(input("number of terms: "))

# partial sums of series of Madelung constant
M = np.zeros(N)

for n in range(N):
    l = int(np.ceil(int(np.sqrt(n+1))))
    for i in range(-l, l + 1):
        for j in range(-l, l + 1):
            for k in range(-l, l + 1):
                if i == 0 and j == 0 and k == 0:
                    continue
                r_square = i ** 2 + j ** 2 + k ** 2
                if r_square <= (n+1):
                    sign = (-1) ** (i + j + k)
                    M[n] += sign / np.sqrt(r_square)

print(f"Sum of first {N} terms is {M[N-1]}")

# plot
x = np.arange(1, N + 1)
plt.figure(figsize=(7, 5))
plt.plot(x, M, label="Partial sums of series")
plt.axhline(y=-1.7476, color="red", linestyle="--", label="Reference -1.7476")
plt.xlabel("N (number of terms)")
plt.ylabel("Convergence behaviour")
plt.title("Madelung constant calculation for NaCl")
plt.grid(True, linestyle="--", alpha=0.5)
plt.xticks(x)
plt.legend()
plt.tight_layout()
plt.show()

```
